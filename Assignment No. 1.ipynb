```python
# Create a dictionary with 5 elements
my_dict = {
    "name": "Yash",
    "rollno": 34,
    "branch": "CSE",
    "year": 1,
    "marks": 85
}

print("Original student information:")
print(my_dict)

# Add new student information
my_dict["college"] = "MIT-WPU"

print("\nAfter adding college information:")
print(my_dict)

# Delete student information
del my_dict["year"]

print("\nAfter deleting year information:")
print(my_dict)

# Display student information
print("\nFinal student information:")
for key, value in my_dict.items():
    print(key, ":", value)
```

```python
string = input("Enter a string: ")

frequency = {}

for char in string:
    if char.isalpha():
        char = char.lower()
        frequency[char] = frequency.get(char, 0) + 1

print("Frequency of letters:")

for char, count in frequency.items():
    print(char, ":", count)
```

```python
numerator = int(input("Enter numerator: "))
denominator = int(input("Enter denominator: "))

if denominator == 0:
    print("Denominator cannot be zero.")
else:
    quotient = numerator // denominator
    remainder = numerator % denominator

    result = (quotient, remainder)

    print("Tuple of Quotient and Remainder:", result)
```
