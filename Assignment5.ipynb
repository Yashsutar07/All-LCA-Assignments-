```python
# Program to check whether a string contains only
# alphabets (a-z, A-Z) and digits (0-9)

import re

# Accept string from user
string = input("Enter a string: ")

# Check the string using regular expression
if re.fullmatch(r'[a-zA-Z0-9]+', string):
    print("The string contains only alphabets and digits.")
else:
    print("The string contains special characters.")
```
